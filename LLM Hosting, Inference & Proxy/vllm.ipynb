```python
import gradio as gr
from vllm import LLM, SamplingParams

# Initialize the model

llm = LLM("meta-llama/Llama-3.2-1B", dtype="half")

# Define the chat function
def chat(input_text):
    sampling_params = SamplingParams(
        temperature=0.7, max_tokens=200
    )
    output = llm.generate(input_text, sampling_params)
    return output[0].outputs[0].text

# Create a Gradio interface
with gr.Blocks() as demo:
    gr.Markdown("## 🤖 Llama-3.2-1B Chatbot")
    chatbox = gr.Textbox(placeholder="Type your message here...")
    outputbox = gr.Textbox(label="Response")

    chatbox.submit(chat, inputs=chatbox, outputs=outputbox)

# Launch the app
demo.launch(debug=True, share=True)
```
